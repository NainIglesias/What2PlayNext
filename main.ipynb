```python
import requests as rq
import numpy as np
import pandas as pd
url = "https://api.steampowered.com/IPlayerService/GetOwnedGames/v0001/"
params = {
    "key": "",
    "steamid": "",
    "include_appinfo": "1",
    "format": "json"
}

data = rq.get(url=url, params=params).json()
games=pd.DataFrame(data["response"]["games"])
#games["playtime_forever"].mean()
games["playtime_forever"] = games["playtime_forever"]/60
most_played_games = games.nlargest(5, "playtime_forever")
```

```python
game_url = "https://store.steampowered.com/api/appdetails"

games_extended_data = []
for appid in most_played_games["appid"]:

    game_params = {
        "appids": appid,
        "l": "spanish",
        "cc": "es"
    }

    response= rq.get(url=game_url, params=game_params).json()
    info = response[str(appid)] 
    
    if info["success"]:
        games_extended_data.append(info["data"])
        

games_extended_data = pd.DataFrame(games_extended_data)

    
```

```python
games_extended_data
games_extended_data["genres"] = games_extended_data["genres"].apply(
    lambda genres : [g["description"] for g in genres] if isinstance(genres, list) else []
)
games_extended_data["genres_ids"] = games_extended_data["genres"].apply(
    lambda genres : [g["description"] for g in genres] if isinstance(genres, list) else []
)
games_extended_data["genres_ids"] = games_extended_data["genres"].apply(
    lambda genres_ids : [g["id"] for g in genres_ids] if isinstance(genres_ids, list) else []
)
games_extended_data["genres_ids"]
```
